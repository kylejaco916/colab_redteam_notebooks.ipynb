```python
# @title 🐙 2. Push All Notebooks to GitHub Repository
# @markdown Enter your GitHub Personal Access Token (PAT) to commit and push all notebooks to GitHub.
# @markdown (Create PAT at https://github.com/settings/tokens with 'repo' scope, or store in Colab Secrets as GITHUB_TOKEN)
import os, subprocess, shutil, base64

try:
    from google.colab import userdata
    token_secret = userdata.get('GITHUB_TOKEN')
except Exception:
    token_secret = ''

token_input = '' # @param {type:'string'}
GITHUB_TOKEN = token_secret if token_secret else token_input.strip()

REPO_OWNER = 'kylejaco916' # @param {type:'string'}
REPO_NAME = 'colab_redteam_notebooks.ipynb' # @param {type:'string'}
BRANCH = 'main' # @param {type:'string'}

if not GITHUB_TOKEN:
    print('⚠️ Please enter your GitHub Personal Access Token in the token field (or add GITHUB_TOKEN to Colab Secrets)!')
else:
    print(f'🚀 Preparing Git push to https://github.com/{REPO_OWNER}/{REPO_NAME}...')
    WORK_DIR = '/content/gh_sync_repo'
    if os.path.exists(WORK_DIR):
        shutil.rmtree(WORK_DIR)
    
    clone_url = f'https://{REPO_OWNER}:{GITHUB_TOKEN}@github.com/{REPO_OWNER}/{REPO_NAME}.git'
    res = subprocess.run(['git', 'clone', clone_url, WORK_DIR], capture_output=True, text=True)
    if res.returncode != 0:
        print('Clone failed or repo empty, initializing fresh git repo...')
        os.makedirs(WORK_DIR, exist_ok=True)
        subprocess.run(['git', 'init', '-b', BRANCH], cwd=WORK_DIR)
        subprocess.run(['git', 'remote', 'add', 'origin', clone_url], cwd=WORK_DIR)
    
    # Ensure payloads are available
    payloads = globals().get('NOTEBOOK_PAYLOADS')
    if payloads:
        for name, b64_content in payloads.items():
            with open(os.path.join(WORK_DIR, name), 'wb') as f:
                f.write(base64.b64decode(b64_content))
    else:
        # Copy from /content
        for f in os.listdir('/content'):
            if f.endswith('.ipynb') or f.endswith('.md') or f.endswith('.py'):
                shutil.copy(os.path.join('/content', f), os.path.join(WORK_DIR, f))
    
    subprocess.run(['git', 'config', 'user.name', REPO_OWNER], cwd=WORK_DIR)
    subprocess.run(['git', 'config', 'user.email', f'71399545+{REPO_OWNER}@users.noreply.github.com'], cwd=WORK_DIR)
    subprocess.run(['git', 'add', '-A'], cwd=WORK_DIR)
    
    diff_res = subprocess.run(['git', 'diff-index', '--quiet', 'HEAD', '--'], cwd=WORK_DIR)
    if diff_res.returncode == 0:
        print('ℹ️ All notebooks on GitHub are already up to date!')
    else:
        subprocess.run(['git', 'commit', '-m', 'Add all verified project Colab notebooks (Red Team suite, AI Stack, ComfyUI, Cloud Bridge)'], cwd=WORK_DIR)
        push_res = subprocess.run(['git', 'push', '-u', 'origin', BRANCH], cwd=WORK_DIR, capture_output=True, text=True)
        if push_res.returncode == 0:
            print(f'🎉 SUCCESS! Pushed all notebooks to https://github.com/{REPO_OWNER}/{REPO_NAME}')
        else:
            print('Push failed. Attempting rebase...')
            subprocess.run(['git', 'pull', 'origin', BRANCH, '--rebase'], cwd=WORK_DIR)
            push_res2 = subprocess.run(['git', 'push', '-u', 'origin', BRANCH], cwd=WORK_DIR, capture_output=True, text=True)
            if push_res2.returncode == 0:
                print(f'🎉 SUCCESS! Pushed all notebooks to https://github.com/{REPO_OWNER}/{REPO_NAME}')
            else:
                print('Push error:', push_res2.stdout, push_res2.stderr)
```

```python
# @title 📦 3. Download All Notebooks as a ZIP Archive
# @markdown Run this cell to download a zip archive of all project notebooks directly to your browser / phone.
import os, zipfile, base64
from google.colab import files

zip_path = '/content/colab_notebook_library.zip'
payloads = globals().get('NOTEBOOK_PAYLOADS')
with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zf:
    if payloads:
        for name, b64_content in payloads.items():
            zf.writestr(name, base64.b64decode(b64_content))
    else:
        for f in os.listdir('/content'):
            if f.endswith('.ipynb') or f.endswith('.md'):
                zf.write(os.path.join('/content', f), arcname=f)

print(f'✅ Created {zip_path} ({os.path.getsize(zip_path)} bytes)')
files.download(zip_path)
```
